```python
## tensor datatypes
# common errors --> inconsistent device types for diff tensors
#               --> inconsistent type
#               --> inconsistent shape
## float
float_tensor = torch.tensor([2.0, 6.0, 3.6],
                             dtype=None, #what data type (float32, float64, float16)
                             device=None,
                             requires_grad=False)
print(f"float tensor: {float_tensor}")
#convert to float 64, use the .to() method
float_64 = float_tensor.to(torch.float64)
print(float_64)
#convert to float 16
float_16 = float_tensor.to(torch.float16)
print(float_16)
print(torch.get_default_dtype())
#inspect a float datatype(finfo), integer(iinfo)
print(torch.finfo(torch.float16))
print(torch.iinfo(torch.int32))

#integer tensor
print(f"===integer type ===================")
integer_tensor = torch.tensor([1,2,4], dtype=None, device=None, requires_grad=False)
print(f"integer tensor: {integer_tensor.dtype}")
#converting to another int type
print(f" int32: {integer_tensor.to(torch.int32)}")

#boolean tensor
print(f"===bolean type =====================")
boolean_tensor = torch.tensor([1,2,3], dtype=torch.bool)
print(f"Boolean tensor: {boolean_tensor}")
# get info (no binfo, only returns True/False)

#complex type (dosent have a half(16), just 32/64 for now)
print(f"=====complex type(less common)========")
complex_tensor = torch.tensor([1+2j, 2+3j], dtype=torch.complex64)
print(f"complex tensor: {complex_tensor}")
print(f"convert to another type: {complex_tensor.to(torch.complex32)}")
#see real and imaginary part
print(f"real part: {complex_tensor.real}, imaginary part: {complex_tensor.imag}")

# type promotion (mixing datatypes in operations)
print(f"Promotion rule: complex > floating > integral > boolean")
# means pytorch promotes the result of an operation of mixed datatypes to the
# highest order in the operation (int + float --> float, float16 + float 32 --> float32)
a = torch.tensor([1,2], dtype=torch.int8)
b = torch.tensor([2,4], dtype=torch.float16)
print(f'====type promotion====')
print(a.dtype)
print(b.dtype)
print(f"result: {a * b}")
c = torch.tensor([[1,3], [2,4]])
print(f"Element size: {c.element_size()}")
print(f"number of elements: {c.numel()}")
print(c.dtype)
print(f"result: {b + c}")
# If you mix a tensor with a raw Python scalar (not a tensor), a separate rule applies — 
# scalars are "weak" and get their type inferred contextually
int_scalar = a + 1.5
print(f" tensor_integer + scalar: {int_scalar.dtype}")
# the result of the above tensor is float32, because the scalar assumes the default float type which is 
# tensor.get_default_dtype() --> float32
print(f"default dtype for scalars: {torch.get_default_dtype()}")

#type_as: used for creating another tensor of the type of another tensor
print("===type_as========")
int_tensor = torch.tensor([1,2], dtype=torch.float32)
print(int_tensor.dtype)
matched_tensor = integer_tensor.type_as(int_tensor)
print(f"matched tensor type: {matched_tensor.dtype}")

## getting the device tensor is running on
getting_Device = torch.tensor([1,2])
print(f"getting device type: {getting_Device.device}")

#return a python number(use .item() on tensor) // concerting a scalar
tensor_type = torch.tensor([1])
print(f"python number from a tensor: {tensor_type.item()}")

```

## difference between elementsize() and numel()
the element size is gotten from the dtype value of the tensor, it is the number of bytes used to store a single element in the tensor, numel() is the number of element in the tensor.

so for a type of 64, if 8 bits make a byte, then 64 / 8 = 8, so elementsize is 8, 
for memory size of a tensor(elementsize() * numel())

```python
## manupulating tensors
## ADDITION
tensor = torch.tensor([1,2,4])
print(f"Addition: {tensor + 10}")
## MULTIPLICATION
print(f"multiplication: {tensor * 2}")
## SUBTRACTION
print(f"sutraction: {tensor - 3}")
## using built in functions
print(f"{torch.add(tensor, 5)}")
print(f"{torch.multiply(tensor, 4)}")
## MATRIX MULTIPLICATION
# LINK: https://www.mathsisfun.com/algebra/matrix-multiplying.html
# LINK: 
#types: Element-wise multiplicaiton / dot-product multiplication
# element wise multiplication of tensors
# element wise
print(f"element wise multiplication: {torch.tensor([1,2,3]) * torch.tensor([4,5,6])}")
new_tensor = torch.tensor([3,4,5])
print(f"matrix multiplication: {tensor @ new_tensor}")
first = torch.rand(1, 3, 4)
second = torch.rand(1, 4, 2)
rprint(f"first: {first.shape}")
rprint(f"second: {second.shape}")
rprint(f"matrix mulitplication result: {first @ second}")
rprint(f"matrix mulitplication result: {torch.matmul(first, second).ndim}")
## rules for matrix multiplication
## 1. inner dimensions must match
# """
# (2, 3) * (2, 3) --> wont work
# (3, 2) * (2, 3) --> work(inner dimension (first(col), second(row))
# """
## 2. the resulting matrix has the shape of the "outer dimensions"
# (2, 3) * (3, 4) --> shape is (2, 4)
print(f"matrix mul: {torch.tensor([2,3]) @ torch.tensor([3,4])}") # --> shape is (2, 4)



```

```python
# shapes for matrix multiplication
tensor_A = torch.tensor([[1,2],
                         [3,4],
                         [5,6]])
tensor_B = torch.tensor([[7,8],
                         [9,10],
                         [11,12]])
# print(f"matrix multiplication: {torch.mm(tensor_A, tensor_B)}")
# above dosent work because of the shape problem, so we transpose to do it
## to transpose a matrix, means to switch the axes
print(tensor_A.T)
# now we can mm
print(f"matrix multiplication: {torch.mm(tensor_A.T, tensor_B)}")

```

```python
## TENSOR AGGREGATION (MIN, MAX, MEAN, SUM)
x = torch.arange(0, 100, 10, dtype=torch.float16)
print(x.type(torch.float32))
print(x.shape)
print(f"minimum: {x.min()}, or minimum: {torch.min(x)}")
print(f"maximum: {x.max()}, or maximum: {torch.max(x)}")
#mean only takes floating numbers(32, 16, 48)
print(f"mean: {x.mean()}, or mean: {torch.mean(x)}")
print(f"sum: {x.sum()}, or sum:{torch.sum(x)}")
# argmin(index of the minimum number)
b = torch.rand(3,4)
print(b)
print(f"argmin: {torch.argmin(b, dim=1, keepdim=True)} or {b.argmin()}")
# argmax:(index of the maximum value in tensor)
print(f"argmax: {x.argmax()} or: {b.argmax()}")
# for a multidemsional tensor
c = torch.rand(4, 3, 2)
print(f"shape: {c.shape}, num dim: {c.ndim}")
rprint(c)
d = torch.argmin(c, dim=1)
print(d, d.shape, d.ndim)
```

```python
## RESHAPING, RESTACKING, SQUEEZING, AND UNSQUEEZING
# reshaping - reshapes an input tensor to a defined shape
# view - Return a view of an inpout tensor of certain shape but keep the same memory as the original tensor
# stacking - combine multiple tensors on top of each other (vstack) or side by side (hstack)
# squeeze - remove all 1 dimensions from a tensor
# unsqueeze - adds a 1 dimension to target tensor
# permute - return a view of the input with dimensions permuted(swapped) in a cetain way
```

```python
x  = torch.arange(1., 10.)
print(x, x.shape)

# add an extra dimension ()must be of same dimension)
x_reshaped = x.reshape(3, 3)
x_reshaped, x_reshaped.shape

# change the view (a view of x, shares the same memory of x)
z = x.view(1, 9)
z, z.shape
# change the value at and index and see
z[:, 0] = 5
z, x
# stack tensors untop of each other
x_stacked = torch.stack([x, x, x], dim=0)
x_stacked
# hstack(side by side) ps: size of tensor must match before stacking
a_tensor = torch.arange(1,7)
h_stacked = torch.hstack((x, a_tensor))
h_stacked, h_stacked.shape
# vstack(untop of each other)
v_stack = torch.vstack((x, x))
v_stack, v_stack.shape
# squeeze - removes all single dimensions from a target tensor (dosent work if dim is not 1)
sq_tensor = torch.rand(1, 3, 1, 4)
print(sq_tensor.shape)
squeezed_tensor = sq_tensor.squeeze(dim=0)
print(f"squeezed_tensor: {squeezed_tensor}")
print(squeezed_tensor.shape)
#unsqueeze --> adds a single dimension to a target tensor at a specific dimension
print(f"squeezed_tensor: {squeezed_tensor}")
unsqueezed_tensor = squeezed_tensor.unsqueeze(dim=0)
print(f"unsqueezed tensor: {unsqueezed_tensor}")
# permute --> raarrage 
x = torch.rand(1,2,2)
print(x)
print(x.size())
#transpose = swap 2 dims. permute = rearrange ALL dims, your way, in one shot. 
print(f"permute tensor: {torch.permute(x, (2, 0 ,1))}")
print(f"attempt to transpose: {x.transpose(0,1)}")


```
