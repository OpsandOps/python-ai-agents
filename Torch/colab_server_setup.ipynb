```python
# Set your authentication token (change this to something secure!)
TOKEN = "my_secure_token_123"

print(f"✅ Your token is: {TOKEN}")
print("⚠️  Keep this token secure - you'll need it to connect from VS Code")
```

```python
# Start Jupyter server
import os
os.environ['TOKEN'] = TOKEN

# Kill any existing jupyter processes
!killall jupyter-notebook 2>/dev/null || true

# Start Jupyter in background
get_ipython().system_raw(
    f'jupyter notebook --ip=0.0.0.0 --port=8888 --no-browser '
    f'--allow-root --NotebookApp.token={TOKEN} '
    f'--NotebookApp.allow_origin="*" > jupyter.log 2>&1 &'
)

print("✅ Jupyter server started!")
print("📄 Check jupyter.log for any errors")
```

```python
# Wait for server to start
import time
print("⏳ Waiting for Jupyter server to initialize...")
time.sleep(5)
print("✅ Server should be ready now")
```

```python
# Create ngrok tunnel
from pyngrok import ngrok

# Close any existing tunnels
ngrok.kill()

# Create new tunnel
public_url = ngrok.connect(8888)
print("\n" + "="*60)
print("🎉 SUCCESS! Your Colab Jupyter server is ready!")
print("="*60)
print(f"\n📍 Public URL: {public_url}")
print(f"🔑 Token: {TOKEN}")
print(f"\n🔗 Full connection URL for VS Code:")
print(f"   {public_url}/?token={TOKEN}")
print("\n" + "="*60)
print("\n📌 Instructions for VS Code:")
print("1. Open your notebook (app.ipynb)")
print("2. Click 'Select Kernel' in the top-right")
print("3. Choose 'Existing Jupyter Server'")
print("4. Paste the full connection URL above")
print("5. Press Enter")
print("\n⚠️  IMPORTANT: Keep this Colab notebook running!")
print("   If you close this tab, the connection will drop.")
```

```python
# Verify GPU access
import torch
print("\n🖥️  System Information:")
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"CUDA version: {torch.version.cuda}")
else:
    print("⚠️  No GPU detected. Did you enable GPU in Runtime > Change runtime type?")
```

```python
# Keep this cell running to maintain the connection
print("\n🟢 Server is active and waiting for connections...")
print("🔄 Press the Stop button to shut down the server")

# This will keep the cell running
try:
    while True:
        time.sleep(60)
except KeyboardInterrupt:
    print("\n🔴 Server stopped")
    ngrok.kill()
```
